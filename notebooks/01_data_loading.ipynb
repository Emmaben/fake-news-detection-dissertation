```python
import pandas as pd
import numpy as np

pd.set_option("display.max_colwidth", 200)
pd.set_option("display.max_columns", None)
```

```python
import os
print("CWD:", os.getcwd())
```

```python
# Load LIAR dataset splits
liar_train = pd.read_csv("../data/liar/train.tsv", sep="\t", header=None)
liar_valid = pd.read_csv("../data/liar/valid.tsv", sep="\t", header=None)
liar_test  = pd.read_csv("../data/liar/test.tsv",  sep="\t", header=None)

print("LIAR shapes:")
print("Train:", liar_train.shape)
print("Valid:", liar_valid.shape)
print("Test :", liar_test.shape)
```

```python
liar_columns = [
    "id", "label", "statement", "subject", "speaker",
    "job", "state", "party",
    "barely_true", "false", "half_true",
    "mostly_true", "pants_on_fire", "context"
]

liar_train.columns = liar_columns
liar_valid.columns = liar_columns
liar_test.columns  = liar_columns
```

```python
liar_train = liar_train[["statement", "label"]]
liar_valid = liar_valid[["statement", "label"]]
liar_test  = liar_test[["statement", "label"]]

liar_train.head()
```

```python
fake_labels = {"false", "barely-true", "pants-on-fire"}

def convert_liar_label(label):
    label = str(label).strip().lower()
    return 0 if label in fake_labels else 1

liar_train["label"] = liar_train["label"].apply(convert_liar_label)
liar_valid["label"] = liar_valid["label"].apply(convert_liar_label)
liar_test["label"]  = liar_test["label"].apply(convert_liar_label)
```

```python
print("LIAR train label counts:\n", liar_train["label"].value_counts(dropna=False))
print("LIAR valid label counts:\n", liar_valid["label"].value_counts(dropna=False))
print("LIAR test  label counts:\n", liar_test["label"].value_counts(dropna=False))
```

```python
print("LIAR Train label distribution:")
print(liar_train["label"].value_counts())

print("\nLIAR Test label distribution:")
print(liar_test["label"].value_counts())
```

```python
# Load ISOT dataset files
fake_news = pd.read_csv("../data/isot/Fake.csv")
real_news = pd.read_csv("../data/isot/True.csv")

print("ISOT raw shapes:")
print("Fake:", fake_news.shape)
print("Real:", real_news.shape)
```

```python
fake_news["label"] = 0
real_news["label"] = 1

isot = pd.concat([fake_news, real_news], axis=0)

# Keep only text + label
isot = isot[["text", "label"]]
isot.columns = ["statement", "label"]

isot.head()
```

```python
isot = isot.sample(frac=1, random_state=42).reset_index(drop=True)
```

```python
print("ISOT label distribution:")
print(isot["label"].value_counts())
```

```python
print("Missing values check:")

print("\nLIAR Train:")
print(liar_train.isnull().sum())

print("\nISOT:")
print(isot.isnull().sum())
```

```python
print("Final dataset summaries:")
print("\nLIAR Train:", liar_train.shape)
print("LIAR Valid:", liar_valid.shape)
print("LIAR Test :", liar_test.shape)
print("ISOT Total:", isot.shape)
```

```python
liar_train.to_csv("../data/liar/liar_train_clean.csv", index=False)
liar_valid.to_csv("../data/liar/liar_valid_clean.csv", index=False)
liar_test.to_csv("../data/liar/liar_test_clean.csv", index=False)

isot.to_csv("../data/isot/isot_clean.csv", index=False)

print("Cleaned datasets saved.")
```
